```python
# matrix product

import numpy as np

# Define two matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Matrix multiplication using @ operator
C = A @ B

#
C = np.matmul(A, B)

print(C)
```

```python
# "element-wise product" or "Hadamard product".
import numpy as np

# Define two arrays
arr1 = np.array([2, 3])
arr2 = np.array([4, 5])

# Element-wise multiplication using *
result = arr1 * arr2
print(result)

# [ 8 15]
```
